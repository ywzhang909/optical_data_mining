# 格式转换

2. 转换时间戳
3. 保存为新文件 

```python
import pandas as pd

from scipy.interpolate import interp1d
```

```python
data = pd.read_csv('data/光轴质心20250606 10：23：07(20%平顶31).TXT', encoding='gbk', header=None, sep='\t')
data.columns = ['time', 'x', 'y']

data['time'] = pd.to_datetime(data['time'], format='%Y-%m-%d-%H：%M：%S.%f')
data.to_csv('data/光轴质心20250606 10：23：07(20%平顶31).csv', index=False)
data
```
