```python
import pandas as pd

from glob import glob
import pathlib
import tqdm

import pygwalker
import matplotlib.pyplot as plt

import numpy as np
from scipy.optimize import curve_fit
```

```python
data = {}
for file in tqdm.tqdm(glob('大气数据/20250611_atmosphere/**/*.xlsx')):
    path = pathlib.Path(file)
    name = path.stem
    df = pd.read_excel(path)

    df.columns = ['time'] + list(df.columns[1:])
    # data[name] = df.set_index('time')
    df['time']='2025-'+df['time']
    df['time'] = pd.to_datetime(df['time'])

    data[name] = df

data.keys()
```
